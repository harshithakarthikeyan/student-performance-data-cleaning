```python
## Objectives of Analysis

'''The following insights are derived from the dataset:

- Analyze the relationship between study hours and student marks
- Examine how attendance impacts academic performance
- Calculate average performance across subjects
- Identify low-performing students based on marks and attendance
- Determine the most challenging subject using score variability'''

### Assumption
# Missing grade values are treated as 0, assuming absent students received a score of 0.









import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("student_performance_dataset.csv")
```

```python
df.info()
```

```python
df.head(10)
```

```python
df.describe()
```

```python
df['Math_Marks']=df['Math_Marks'].fillna(0)
```

```python
df['Science_Marks']=df['Science_Marks'].fillna(0)
```

```python
df['English_Marks']=df['English_Marks'].fillna(0)
```

```python
df['Attendance_Percent']=df['Attendance_Percent'].fillna(df['Attendance_Percent'].mean().round(2))
```

```python
df['Study_Hours']=df['Study_Hours'].fillna(df['Study_Hours'].mean().round(2))
```

```python
df["Average_Marks"] = df[["Math_Marks", "Science_Marks", "English_Marks"]].mean(axis=1).round(2)
```

```python
study=pd.to_numeric(df['Study_Hours'],errors='coerce').to_numpy()
avg_marks = df["Average_Marks"].to_numpy()
np.corrcoef(study, avg_marks)
```

**Inference:** The correlation value (~ -0.13) indicates a very weak relationship between study hours and average marks.

```python
attendance=df['Attendance_Percent'].to_numpy()
```

```python
np.corrcoef(avg_marks,attendance)
```

**Inference:** Attendance shows almost no relationship with overall student performance.

```python
df.head(10)
```

```python
df[["Math_Marks", "Science_Marks", "English_Marks"]].mean()
```

**Inference:** The average marks vary across subjects, indicating differences in subject difficulty.


```python
overall_avg = df["Average_Marks"].mean()
overall_avg
```

```python
low_performers = df[df["Average_Marks"] < overall_avg]
low_performers
```

**Inference:** Students scoring below the overall average were identified as low-performing.
